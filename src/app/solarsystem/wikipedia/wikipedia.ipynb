```python
from wikiScraper import fetch_wikipedia, save_wikipedia
import os

planets = ["Mercury_(planet)", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]
for planet in planets:
    directory = f"./Planets/{planet}"
    os.makedirs(directory, exist_ok=True)
    save_wikipedia(planet, directory=directory)
```

```python
more = {
    "Mercury_(planet)": ["Atmosphere_of_Mercury"], 
    "Venus": ["Atmosphere_of_Venus"], 
    "Earth": ["Internal_structure_of_Earth", "Atmosphere_of_Earth"],
    "Mars": ["Atmosphere_of_Mars"],
    "Jupiter": ["Atmosphere_of_Jupiter"],
    "Uranus": ["Atmosphere_of_Uranus"],
}
for planet, values in more.items():
    for topic in values:
        directory = f"./Planets/{planet}/"
        os.makedirs(directory, exist_ok=True)
        save_wikipedia(topic, directory=directory)
```
